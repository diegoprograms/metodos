```python
# Programa sencillo sobre contaminación atmosférica

print("ANÁLISIS DE CONTAMINACIÓN ATMOSFÉRICA")
print("--------------------------------------")

# Mediciones de PM2.5 en µg/m³
pm25 = [15, 22, 35, 40, 28]

# Calcular el promedio
promedio = sum(pm25) / len(pm25)

print("Mediciones de PM2.5:", pm25)
print("Promedio de PM2.5:", promedio, "µg/m³")

# Evaluar el nivel de contaminación
if promedio <= 15:
    print("Nivel: Bueno")
elif promedio <= 35:
    print("Nivel: Moderado")
else:
    print("Nivel: Alto")
```
