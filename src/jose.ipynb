```python
# MONITOREO AMBIENTAL DE UN RÍO

print("====================================")
print("   MONITOREO AMBIENTAL DE UN RÍO")
print("====================================")

# Ingreso de datos
ph = float(input("Ingrese el pH del agua: "))
nitratos = float(input("Ingrese los nitratos (mg/L): "))
fosfatos = float(input("Ingrese los fosfatos (mg/L): "))
turbidez = float(input("Ingrese la turbidez (NTU): "))
temperatura = float(input("Ingrese la temperatura (°C): "))

print("\n========== RESULTADOS ==========")

# Evaluación del pH
if 6.5 <= ph <= 8.5:
    print("pH: Normal")
else:
    print("pH: Fuera del rango de referencia")

# Evaluación de nutrientes
if nitratos <= 10:
    print("Nitratos: Aceptables")
else:
    print("Nitratos: Elevados")

if fosfatos <= 0.1:
    print("Fosfatos: Aceptables")
else:
    print("Fosfatos: Elevados")

# Evaluación de turbidez
if turbidez <= 5:
    print("Turbidez: Baja")
else:
    print("Turbidez: Elevada")

# Evaluación de temperatura
if 10 <= temperatura <= 25:
    print("Temperatura: Dentro del rango de referencia")
else:
    print("Temperatura: Fuera del rango de referencia")

print("====================================")
```
